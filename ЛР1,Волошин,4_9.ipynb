### Завдання 25
Дано два вектори:

```text

b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
# Код студента
a = np.array([1, 2])
b = np.array([2, 1])

cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle_deg = np.degrees(np.arccos(cos_theta))

print("Косинус кута:", cos_theta)
print("Кут у градусах:", angle_deg)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]])

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)
Z = (A - mu) / sigma

print("Нормалізована матриця:\n", Z)
````

X = [[2, 4, 6],

Обчисліть **евклідову відстань** між:

- другим і третім рядком.

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])
b = np.array([9, 8, 3])

x_sol = np.linalg.solve(A, b)
print("1. Розв'язок системи (x, y, z):", x_sol)

check_b = A @ x_sol
print("2. Перевірка матричним множенням (A @ x):", check_b)
print("3. Вектор b:", b)
print("Збігаються?", np.allclose(check_b, b))
````
